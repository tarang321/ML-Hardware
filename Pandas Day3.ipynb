```python
import numpy as np
import matplotlib.pyplot as plt
def f(t):
    return np.exp(-t) * np.cos(2*np.pi*t)
t1 = np.arange(0.0, 5.0, 0.1)
print(t1)
t2 = np.arange(0.0, 5.0, 0.002)
plt.figure(1)
plt.plot(t1, f(t1), 'r*', t2, f(t2), "g")
```

```python

import matplotlib.pyplot as plt
labels = 'Frogs', 'Hogs', 'Dogs', 'Logs'
sizes = [15, 30, 25, 30]
explode = (0, 0.2, 0, 0)  		# only "explode" the 2nd slice (i.e. 'Hogs')

fig1, ax1 = plt.subplots()
ax1.pie(sizes, explode=explode, labels=labels, autopct='%1.1f%%',
        shadow=True, startangle=40)
ax1.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python

mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)
# the histogram of the data
#plt.hist?
n, bins, patches = plt.hist(x, 80, density=.5)
plt.xlabel('Smarts')
plt.ylabel('Probability')
plt.title('Histogram of IQ')
plt.text(60, .025, r'$\mu=100,\ \sigma=15$')
plt.axis([40, 160, 0, 0.03])
plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
n = np.array([0,1,2,3,4,5])
xx = np.linspace(-0.75, 1., 100)
fig, axes = plt.subplots(1, 4, figsize=(14,4))
axes[0].scatter(xx, xx + 0.25*np.random.randn(len(xx)))
axes[0].set_title("scatter")

axes[1].step(n, n**2, lw=4)		
axes[1].set_title("step")
axes[2].bar(n, n**2, align="center", width=1, alpha=1)
axes[2].set_title("bar")
axes[3].fill_between(xx, xx**2,  color="red", alpha=1);
axes[3].set_title("fill_between");
plt.savefig("graphs.pdf",dpi=5000, format="pdf")
```
